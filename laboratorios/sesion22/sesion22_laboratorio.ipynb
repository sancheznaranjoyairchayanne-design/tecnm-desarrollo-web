### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from sqlalchemy import create_engine, String, Integer, ForeignKey
from sqlalchemy.orm import (
    DeclarativeBase,
    Mapped,
    mapped_column,
    relationship,
    sessionmaker
)

# Base de datos SQLite
engine = create_engine(
    "sqlite:///biblioteca.db",
    connect_args={"check_same_thread": False}
)

SessionLocal = sessionmaker(bind=engine)


class Base(DeclarativeBase):
    pass


# ==========================
# Modelos ORM
# ==========================

class Autor(Base):
    __tablename__ = "autores"

    id: Mapped[int] = mapped_column(Integer, primary_key=True)
    nombre: Mapped[str] = mapped_column(String(100))

    libros: Mapped[list["Libro"]] = relationship(
        back_populates="autor"
    )


class Libro(Base):
    __tablename__ = "libros"

    id: Mapped[int] = mapped_column(Integer, primary_key=True)
    titulo: Mapped[str] = mapped_column(String(100))
    anio: Mapped[int] = mapped_column(Integer)

    autor_id: Mapped[int] = mapped_column(
        ForeignKey("autores.id")
    )

    autor: Mapped["Autor"] = relationship(
        back_populates="libros"
    )


Base.metadata.create_all(engine)


# ==========================
# Esquemas Pydantic
# ==========================

class AutorCreate(BaseModel):
    nombre: str


class LibroCreate(BaseModel):
    titulo: str
    anio: int


# ==========================
# API
# ==========================

app = FastAPI(
    title="Tu relación autor → libros se verá así"
)


@app.post("/api/autores", status_code=201)
def crear_autor(datos: AutorCreate):
    db = SessionLocal()

    autor = Autor(nombre=datos.nombre)

    db.add(autor)
    db.commit()
    db.refresh(autor)

    resultado = {
        "id": autor.id,
        "nombre": autor.nombre
    }

    db.close()
    return resultado


@app.post("/api/autores/{autor_id}/libros", status_code=201)
def crear_libro(autor_id: int, datos: LibroCreate):
    db = SessionLocal()

    autor = db.get(Autor, autor_id)

    if not autor:
        db.close()
        raise HTTPException(
            status_code=404,
            detail="Autor no encontrado"
        )

    libro = Libro(
        titulo=datos.titulo,
        anio=datos.anio,
        autor_id=autor_id
    )

    db.add(libro)
    db.commit()
    db.refresh(libro)

    resultado = {
        "id": libro.id,
        "titulo": libro.titulo,
        "anio": libro.anio,
        "autor_id": libro.autor_id
    }

    db.close()
    return resultado


@app.get("/api/autores/{autor_id}")
def obtener_autor(autor_id: int):
    db = SessionLocal()

    autor = db.get(Autor, autor_id)

    if not autor:
        db.close()
        raise HTTPException(
            status_code=404,
            detail="Autor no encontrado"
        )

    resultado = {
        "id": autor.id,
        "nombre": autor.nombre,
        "libros": [
            {
                "id": libro.id,
                "titulo": libro.titulo
            }
            for libro in autor.libros
        ]
    }

    db.close()
    return resultado


@app.get("/api/libros")
def listar_libros():
    db = SessionLocal()

    libros = db.query(Libro).all()

    resultado = {
        "libros": [
            {
                "id": libro.id,
                "titulo": libro.titulo
            }
            for libro in libros
        ]
    }

    db.close()
    return resultado
````
