> **La pregunta de hoy:** ¿Por qué el estándar OAuth2 usa un formulario y no JSON para el login? = OAuth2 Password Flow usa application/x-www-form-urlencoded en lugar de JSON porque así lo define el estándar OAuth 2.0 para la solicitud de obtención de tokens. Esto permite que clientes, servidores y herramientas compatibles con OAuth2 se comuniquen utilizando un formato común y estandarizado.

### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI, Depends, HTTPException, status
from fastapi.security import OAuth2PasswordBearer, OAuth2PasswordRequestForm

app = FastAPI(title="Tu OAuth2 Password Flow se verá así")

oauth2_scheme = OAuth2PasswordBearer(tokenUrl="token")

USUARIO = "ana"
PASSWORD = "secreta"
TOKEN = "token-ana"


@app.post("/token")
def token(form: OAuth2PasswordRequestForm = Depends()):
    if form.username == USUARIO and form.password == PASSWORD:
        return {
            "access_token": TOKEN,
            "token_type": "bearer"
        }

    raise HTTPException(
        status_code=status.HTTP_401_UNAUTHORIZED,
        detail="Credenciales incorrectas",
        headers={"WWW-Authenticate": "Bearer"},
    )


@app.get("/api/perfil")
def perfil(token: str = Depends(oauth2_scheme)):
    if token != TOKEN:
        raise HTTPException(
            status_code=status.HTTP_401_UNAUTHORIZED,
            detail="Token inválido",
            headers={"WWW-Authenticate": "Bearer"},
        )

    return {"usuario": USUARIO}
````
