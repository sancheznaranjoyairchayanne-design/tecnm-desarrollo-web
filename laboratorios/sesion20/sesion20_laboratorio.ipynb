### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from datetime import datetime, timedelta

from fastapi import FastAPI, HTTPException, Header, Depends
from pydantic import BaseModel
from passlib.context import CryptContext
from jose import jwt, JWTError

app = FastAPI(title="Tu Hito 2 completo se verá así")

SECRET_KEY = "hito2-secret-key"
ALGORITHM = "HS256"

pwd_context = CryptContext(
    schemes=["bcrypt"],
    deprecated="auto"
)


class RegistroRequest(BaseModel):
    usuario: str
    password: str


class LoginRequest(BaseModel):
    usuario: str
    password: str


usuarios = {
    "admin": {
        "usuario": "admin",
        "password_hash": pwd_context.hash("admin123"),
        "rol": "admin"
    }
}


def crear_hash(password: str) -> str:
    return pwd_context.hash(password)


def verificar_password(password: str, password_hash: str) -> bool:
    return pwd_context.verify(password, password_hash)


def crear_jwt(usuario: str, rol: str) -> str:
    payload = {
        "sub": usuario,
        "rol": rol,
        "exp": datetime.utcnow() + timedelta(hours=1)
    }

    return jwt.encode(
        payload,
        SECRET_KEY,
        algorithm=ALGORITHM
    )


def obtener_usuario_actual(
    authorization: str = Header(default=None)
):
    if not authorization:
        raise HTTPException(
            status_code=401,
            detail="No autenticado"
        )

    partes = authorization.split()

    if len(partes) != 2 or partes[0].lower() != "bearer":
        raise HTTPException(
            status_code=401,
            detail="Token inválido"
        )

    token = partes[1]

    try:
        payload = jwt.decode(
            token,
            SECRET_KEY,
            algorithms=[ALGORITHM]
        )

        return payload

    except JWTError:
        raise HTTPException(
            status_code=401,
            detail="Token inválido"
        )


def require_admin(
    usuario=Depends(obtener_usuario_actual)
):
    if usuario["rol"] != "admin":
        raise HTTPException(
            status_code=403,
            detail="Acceso denegado"
        )

    return usuario


@app.post("/api/registro", status_code=201)
def registro(datos: RegistroRequest):

    if datos.usuario in usuarios:
        raise HTTPException(
            status_code=400,
            detail="Usuario ya existe"
        )

    usuarios[datos.usuario] = {
        "usuario": datos.usuario,
        "password_hash": crear_hash(datos.password),
        "rol": "user"
    }

    return {
        "usuario": datos.usuario,
        "rol": "user"
    }


@app.post("/api/token")
def login(datos: LoginRequest):

    usuario = usuarios.get(datos.usuario)

    if usuario is None:
        raise HTTPException(
            status_code=401,
            detail="Credenciales inválidas"
        )

    if not verificar_password(
        datos.password,
        usuario["password_hash"]
    ):
        raise HTTPException(
            status_code=401,
            detail="Credenciales inválidas"
        )

    token = crear_jwt(
        usuario["usuario"],
        usuario["rol"]
    )

    return {
        "access_token": token,
        "token_type": "bearer"
    }


@app.get("/api/perfil")
def perfil(
    usuario=Depends(obtener_usuario_actual)
):
    return {
        "usuario": usuario["sub"],
        "rol": usuario["rol"]
    }


@app.get("/api/admin/panel")
def panel_admin(
    admin=Depends(require_admin)
):
    return {
        "panel": "admin"
    }
````
