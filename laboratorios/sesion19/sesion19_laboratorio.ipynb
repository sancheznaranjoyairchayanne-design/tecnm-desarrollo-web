### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI
from pydantic import BaseModel
import html

app = FastAPI(title="Tu API neutralizará el HTML peligroso")

comentarios = []


class Comentario(BaseModel):
    texto: str


@app.post("/api/comentarios", status_code=201)
def crear_comentario(datos: Comentario):
    texto_seguro = html.escape(datos.texto)

    comentario = {
        "id": len(comentarios) + 1,
        "texto": texto_seguro
    }

    comentarios.append(comentario)
    return comentario


@app.get("/api/comentarios")
def listar_comentarios():
    return {"comentarios": comentarios}
````
