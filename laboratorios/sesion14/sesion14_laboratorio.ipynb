### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py

from fastapi import FastAPI, HTTPException, status
from pydantic import BaseModel
import bcrypt

app = FastAPI(title="Tu registro y login se verán así")

# Base de datos temporal en memoria
usuarios = {}


class Credenciales(BaseModel):
    usuario: str
    password: str


@app.post("/api/registro", status_code=status.HTTP_201_CREATED)
def registro(datos: Credenciales):

    if datos.usuario in usuarios:
        raise HTTPException(
            status_code=409,
            detail="El usuario ya existe"
        )

    hash_password = bcrypt.hashpw(
        datos.password.encode(),
        bcrypt.gensalt()
    ).decode()

    usuarios[datos.usuario] = hash_password

    return {
        "usuario": datos.usuario
    }


@app.post("/api/login")
def login(datos: Credenciales):

    hash_guardado = usuarios.get(datos.usuario)

    if hash_guardado is None:
        raise HTTPException(
            status_code=401,
            detail="credenciales inválidas"
        )

    if not bcrypt.checkpw(
        datos.password.encode(),
        hash_guardado.encode()
    ):
        raise HTTPException(
            status_code=401,
            detail="credenciales inválidas"
        )

    return {
        "usuario": datos.usuario,
        "autenticado": True
    }
````
