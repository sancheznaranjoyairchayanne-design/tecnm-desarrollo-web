### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI, HTTPException, Response, status
from pydantic import BaseModel, Field

app = FastAPI(title="Tu Mock API del Hito 1 se comportará así")


class Libro(BaseModel):
    titulo: str = Field(..., min_length=1)
    anio: int = Field(..., ge=1450)


class LibroRespuesta(Libro):
    id: int


libros = {}
siguiente_id = 1


@app.get("/api/libros")
def listar_libros():
    return {"libros": list(libros.values())}


@app.post("/api/libros", response_model=LibroRespuesta, status_code=201)
def crear_libro(libro: Libro):
    global siguiente_id

    nuevo = LibroRespuesta(
        id=siguiente_id,
        titulo=libro.titulo,
        anio=libro.anio
    )

    libros[siguiente_id] = nuevo
    siguiente_id += 1

    return nuevo


@app.get("/api/libros/{libro_id}", response_model=LibroRespuesta)
def obtener_libro(libro_id: int):
    if libro_id not in libros:
        raise HTTPException(
            status_code=404,
            detail="Libro no encontrado"
        )

    return libros[libro_id]


@app.put("/api/libros/{libro_id}", response_model=LibroRespuesta)
def actualizar_libro(libro_id: int, libro: Libro):
    if libro_id not in libros:
        raise HTTPException(
            status_code=404,
            detail="Libro no encontrado"
        )

    actualizado = LibroRespuesta(
        id=libro_id,
        titulo=libro.titulo,
        anio=libro.anio
    )

    libros[libro_id] = actualizado
    return actualizado


@app.delete("/api/libros/{libro_id}", status_code=204)
def eliminar_libro(libro_id: int):
    if libro_id not in libros:
        raise HTTPException(
            status_code=404,
            detail="Libro no encontrado"
        )

    del libros[libro_id]
    return Response(status_code=status.HTTP_204_NO_CONTENT)
````
