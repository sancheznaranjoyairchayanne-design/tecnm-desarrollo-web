> **La pregunta de hoy:** ¿Cómo recuerda el servidor que ya iniciaste sesión? = En una sesión stateful el servidor guarda la información de la sesión y el cliente solo envía un identificador (session_id). En un sistema stateless con JWT, el servidor no guarda la sesión; el token contiene la información necesaria para identificar al usuario.

```http

POST /api/logout  (con la cookie)

GET /api/perfil   (después del logout)
HTTP/1.1 401 Unauthorized
```

_La cookie viaja sola: el cliente no tiene que copiar el identificador en cada petición._

### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py

from fastapi import FastAPI, Response, Cookie, HTTPException
from pydantic import BaseModel
from uuid import uuid4

app = FastAPI(title="Tu flujo de sesión se verá así")

# Usuario de prueba
USUARIO = {
    "usuario": "ana",
    "password": "secreta"
}

# Almacenamiento de sesiones en memoria
sesiones = {}


class Login(BaseModel):
    usuario: str
    password: str


@app.post("/api/login")
def login(datos: Login, response: Response):
    if (
        datos.usuario != USUARIO["usuario"]
        or datos.password != USUARIO["password"]
    ):
        raise HTTPException(
            status_code=401,
            detail="credenciales inválidas"
        )

    session_id = str(uuid4())
    sesiones[session_id] = datos.usuario

    response.set_cookie(
        key="session_id",
        value=session_id,
        httponly=True,
        samesite="lax"
    )

    return {"mensaje": "sesión iniciada"}


@app.get("/api/perfil")
def perfil(session_id: str | None = Cookie(default=None)):
    usuario = sesiones.get(session_id)

    if usuario is None:
        raise HTTPException(
            status_code=401,
            detail="no autenticado"
        )

    return {"usuario": usuario}


@app.post("/api/logout", status_code=204)
def logout(
    response: Response,
    session_id: str | None = Cookie(default=None)
):
    if session_id:
        sesiones.pop(session_id, None)

    response.delete_cookie("session_id")

    return Response(status_code=204)
````
